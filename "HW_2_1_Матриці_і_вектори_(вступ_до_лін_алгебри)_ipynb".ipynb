### Завдання 3: Множення матриць
1. Задані дві матриці `A` і `B`розмірностями 3x2 та 2х3. Виконайте їхнє множення, щоб отримати матрицю 2x2. Виведіть результуючу матрицю `result_matrix` на екран.

2. Заданий вектор `C` розмірністю 2х1. Подумайте спочатку, яку розмірність Ви отрмаєте, коли домножите `result_matrix` на вектор `C` і який результат буде. А потім виконайте множення і виведіть результат на екран.

3. Задана матриця `D` розмірністю 2х2. Подумайте спочатку, яку розмірність Ви отрмаєте, коли домножите `result_matrix` на матрицю `D` і який результат буде. А потім виконайте множення і виведіть результат на екран.

```python
A = np.array(
    [[0.18, 0.53],
     [0.75, 0.47],
     [0.35, 0.21]])

B = np.array(
    [[0.17, 0.58, 0.75],
     [0.38, 0.11, 0.15]])

C = np.array([[0], [1]])

D = np.array([[0,1], [1,0]])

display(A, B, C, D)
```

```python
# B має розмір 2x3, а A — 3x2, тому множимо B на A
result_matrix = B @ A
print('\nРезультат множення B на A:')
print(result_matrix)
print('Розмір:', result_matrix.shape)

# При множенні на C отримаємо вектор 2x1.
# Він буде містити другий стовпець result_matrix.
result_C = result_matrix @ C
print('\nРезультат множення на C:')
print(result_C)
print('Розмір:', result_C.shape)

# При множенні на D отримаємо матрицю 2x2.
# Перший і другий стовпці поміняються місцями.
result_D = result_matrix @ D
print('\nРезультат множення на D:')
print(result_D)
print('Розмір:', result_D.shape)
```

**Висновок**: Щоб отримати матрицю розміром 2×2, потрібно помножити B на A. Після множення на C отримали вектор 2×1 зі значеннями 0.5202 та 0.2846, тобто другий стовпець матриці. Після множення на D розмір матриці залишився 2×2, але її стовпці помінялися місцями.


### Завдання 4: Транспонування матриць
1. Транспонуйте `result_matrix`.
2. Зробіть транспонування вектора `vector` і виведіть форму нової структури. Чи бачите ви зміну?
3. Використайте операцію `numpy.expand_dims` над вектором `vector` з аргументом axis=1. Запишіть отриманий результат в змінну `column_vector` і виведіть на екран. Порівняйте з `vector`.
4. Зробіть транспонування `column_vector`, виведіть результат на екран разом з розмірностями отриманої структури.




```python
print('Транспонована матриця:')
print(result_matrix.T)
print('Розмір:', result_matrix.T.shape)

print('\nТранспонований вектор:', vector.T)
print('Розмір до транспонування:', vector.shape)
print('Розмір після транспонування:', vector.T.shape)

column_vector = np.expand_dims(vector, axis=1)
print('\nВектор-стовпець:')
print(column_vector)
print('Розмір вектора-стовпця:', column_vector.shape)

print('\nТранспонований вектор-стовпець:')
print(column_vector.T)
print('Розмір:', column_vector.T.shape)
```

**Висновок**: Після транспонування матриці її рядки стали стовпцями, але розмір залишився 2×2. При транспонуванні одновимірного вектора його форма (10,) не змінилася. За допомогою np.expand_dims отримано вектор-стовпець розміром (10, 1). Після його транспонування отримано вектор-рядок розміром (1, 10).
