1. Задана функція залежності оцінки від кількості годин, витрачених на навчання: $ f(x) = 10x - 5 $.
  1. З допомогою Python обчисліть похідну цієї функції.
  2. Використовуйте бібліотеку `matplotlib` для побудови графіків функції та її похідної в діапазоні значень від 0 до 10 годин. Проаналізуйте, як зміна кількості годин вивчення впливає на оцінку.



```python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = sp.symbols('x')
f = 10*x - 5
df = sp.diff(f, x)

print('Функція:', f)
print('Похідна:', df)

x_values = np.linspace(0, 10, 100)
y_values = 10*x_values - 5
dy_values = np.full_like(x_values, float(df))

plt.plot(x_values, y_values, label='f(x) = 10x - 5')
plt.plot(x_values, dy_values, label="f'(x) = 10")
plt.xlabel('Години навчання')
plt.ylabel('Значення')
plt.title('Функція оцінки та її похідна')
plt.legend()
plt.grid()
plt.show()
```

**Висновок**: На графіку видно, що зі збільшенням кількості годин навчання оцінка зростає рівномірно. Похідна функції дорівнює 10 і не змінюється. Це означає, що кожна додаткова година навчання збільшує оцінку на 10 балів за заданою моделлю. Наприклад, за 10 годин навчання оцінка становитиме 95 балів.

2. Використовуючи Python і `SymPy`, обчисліть похідні для функцій $ f(x) = \sin(x) $ та $ f(x) = \ln(x) $. Зобразіть ці функції та їхні похідні за допомогою `matplotlib`.



```python
f1 = sp.sin(x)
f2 = sp.log(x)

df1 = sp.diff(f1, x)
df2 = sp.diff(f2, x)

print('Похідна sin(x):', df1)
print('Похідна ln(x):', df2)

x_sin = np.linspace(-2*np.pi, 2*np.pi, 200)

plt.plot(x_sin, np.sin(x_sin), label='sin(x)')
plt.plot(x_sin, np.cos(x_sin), label="sin'(x) = cos(x)")
plt.xlabel('x')
plt.ylabel('Значення')
plt.title('Синус та його похідна')
plt.legend()
plt.grid()
plt.show()

# Логарифмік визначений тільки для x > 0
x_log = np.linspace(0.1, 10, 200)

plt.plot(x_log, np.log(x_log), label='ln(x)')
plt.plot(x_log, 1/x_log, label="ln'(x) = 1/x")
plt.xlabel('x')
plt.ylabel('Значення')
plt.title('Логарифм та його похідна')
plt.legend()
plt.grid()
plt.show()
```

**Висновок**: Похідною синуса є косинус, тому обидва графіки мають хвилеподібну форму, але зміщені один відносно одного. Коли косинус додатний, синус зростає, а коли від’ємний — спадає. Похідна натурального логарифма дорівнює 1/x і є додатною для всіх x > 0. На графіку видно, що логарифм зростає, але зі збільшенням x швидкість його зростання поступово зменшується.


3. Загальні витрати компанії від обсягу виробництва задані функцією $ f(x) = x^2 - 5x + 1 $, де $ x $ — кількість вироблених одиниць. Обчисліть похідну цієї функції з допомогою SymPy та візуалізуйте значення функції і її похідної на відрізку від 0 до 100.

```python
f = x**2 - 5*x + 1
df = sp.diff(f, x)

print('Функція витрат:', f)
print('Похідна:', df)
print('Точка, у якій похідна дорівнює нулю:', float(sp.solve(df, x)[0]))

x_values = np.linspace(0, 100, 200)
y_values = x_values**2 - 5*x_values + 1
dy_values = 2*x_values - 5

plt.figure(figsize=(8, 4))
plt.plot(x_values, y_values, label='f(x) = x² - 5x + 1')
plt.xlabel('Кількість вироблених одиниць')
plt.ylabel('Витрати')
plt.title('Функція загальних витрат')
plt.legend()
plt.grid()
plt.show()

plt.figure(figsize=(8, 4))
plt.plot(x_values, dy_values, label="f'(x) = 2x - 5")
plt.xlabel('Кількість вироблених одиниць')
plt.ylabel('Значення похідної')
plt.title('Похідна функції витрат')
plt.legend()
plt.grid()
plt.show()
```

**Висновок**: Похідна функції витрат дорівнює 2x − 5. Спочатку функція витрат спадає, а при x = 2,5 похідна дорівнює нулю, що відповідає мінімуму функції на неперервному проміжку. Після цієї точки витрати зростають. На графіку похідної видно, що зі збільшенням обсягу виробництва швидкість зростання витрат також збільшується.
